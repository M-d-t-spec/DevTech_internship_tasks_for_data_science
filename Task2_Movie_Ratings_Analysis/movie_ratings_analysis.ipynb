```python
# 🎥 Movie Ratings Analysis (CSV version)
# By Mahlet

import pandas as pd
import matplotlib.pyplot as plt

# Step 1: Read our movie data from the CSV
data = pd.read_csv("movies.csv")  # Make sure the file is uploaded in the same folder

# Step 2: Show the data
print("🎬 Movie Ratings Data:")
print(data)

# Step 3: Group and summarize
avg_rating = data.groupby("Genre")["Rating"].mean().round(2)
count_by_genre = data["Genre"].value_counts()

print("\n⭐ Average rating per genre:")
print(avg_rating)
print("\n🍿 Number of movies per genre:")
print(count_by_genre)

# Step 4: Bar chart — average rating per genre
plt.figure(figsize=(8,5))
avg_rating.plot(kind='bar')
plt.title("🎥 Average Ratings by Genre")
plt.xlabel("Genre")
plt.ylabel("Average Rating")
plt.grid(axis='y')
plt.show()

# 🌟 Step 5: Chart — number of movies per genre
plt.figure(figsize=(8,5))
count_by_genre.plot(kind='bar')
plt.title("🍿 Number of Movies per Genre")
plt.xlabel("Genre")
plt.ylabel("Count")
plt.grid(axis='y')
plt.show()
```

```python
# Trend Insights 📈

# Top 3 highest-rated genres
top_genres = avg_rating.sort_values(ascending=False).head(3)

# Bottom 3 lowest-rated genres
bottom_genres = avg_rating.sort_values().head(3)

print("\n🏆 Top 3 genres by average rating:")
print(top_genres)

print("\n📉 Bottom 3 genres by average rating:")
print(bottom_genres)
```

```python
# Most popular genre by number of movies
most_popular = count_by_genre.idxmax()
most_popular_count = count_by_genre.max()

print(f"\n🍿 Most popular genre: {most_popular} ({most_popular_count} movies)")
```

```python
print("\n🔍 Insights:")
print(f"- '{top_genres.index[0]}' is the highest-rated genre, meaning viewers enjoy it most.")
print(f"- '{most_popular}' has the most movies, showing it’s the most produced or common genre.")
print("- Comparing both helps us understand whether popularity also means higher quality.")
```
